```python
dummy_data = {
    'kg' : '78',
    'hg' : '1.75'
}

validation_limits = {
    'kg':{ 'min': 0, 'max': 300 },
    'hg':{ 'min': 0, 'max': 3 }
}


def convert_to_float( value ):

  try:
    data = float( value )
  except ValueError:
    data = False

  return data

def get_imc_range( user_imc ):

  if user_imc < 18.5:
    range = 'Bajo peso'
  elif user_imc < 25:
    range = 'Peso normal'
  elif user_imc < 30:
    range = 'Sobrepeso'
  else:
    range = 'Obesidad'

  return range


def is_data_ok( unit, value ):

  return ( value < validation_limits[unit]['max'] ) and ( value > validation_limits[unit]['min'])

def get_data( field ):

  if dummy == True:
    print(f"Dummy data! {dummy_data[field]}")
    data = dummy_data[field]
  else:
    data = input()

  return data

def get_user_data( field ):

  if field == 'kg':
    print("Introduce tu peso en kg (ej. 75.5): ")
  else:
    print("Introduce tu altura en metros (ej. 1.75): ")

  return convert_to_float( get_data( field ) )


def request_data( field ):

  while True:
    data = get_user_data( field )
    if data and is_data_ok( field, data ):
      break
    else:
      print("Algo no fue bien, el valor no es correcto.")

  return data


def get_calculate_imc( user_data ):

  return round( user_data['kg'] / pow( user_data['hg'], 2), 2 )


def imc_calculator():

  checked_data = {}
  system_data = ['kg','hg']

  print("Calculadora de indice de masa corporal\n")

  for field in system_data:
      checked_data[field] = request_data( field )

  imc   = get_calculate_imc( checked_data )
  range = get_imc_range( imc )

  print(f"Tu imc es: {imc}, se considera: {range} ")


dummy = True
imc_calculator()
```

```python
def reverse_diccionary( data ):

  reversed_dict = {}
  duplicated_keys = []

  for key,value in data.items():

    if reversed_dict.get(value) == None:
      reversed_dict[value] = key
    elif isinstance( reversed_dict[value], list ):
       reversed_dict[value].append( key )
    else:
      reversed_dict[value] = [reversed_dict[value], key]
      duplicated_keys.append(value)

  if duplicated_keys:
    print(f"Atención! Los siguientes valores del diccionario original que no son únicos: {duplicated_keys}")

  return reversed_dict

survey_1 = {'a':1,'b':2,'c':3}
survey_2 = {'x':'apple', 'y':'banana','z':'banana', 'g': 'apple'}

reverse_diccionary( survey_1 )
reverse_diccionary( survey_2 )
```

```python
def get_file( name, type ="text" ):
  try:
    with open(f"/content/{name}", "rt", encoding="utf-8") as f:
      return f.read() if type == "text" else f.readlines()
  except FileNotFoundError:
    print("No se encontro el fichero!")
    return None

def counter_sort_text_words( text ):

  if text == None:
    return None

  words_counted = get_words_counted( sorted(procces_phrase_to_list( clean_data( text ) )) )

  final_dict = {}
  for word, value in words_counted.items():
    if final_dict.get(word[0]) == None:
      final_dict[word[0]] = { word:value }
    else:
      final_dict[word[0]].update( {word:value} )

  return final_dict


text = get_file("tu me quieres blanca.txt")
counter_sort_text_words( text )
```

```python
entry_list = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

def safe_in_list( org_data, float_data, result_list ):

  if float_data:
    result_list[0].append( float_data )
  else:
    result_list[1].append( org_data )

  return result_list

def walk_list( value_list ):

  if not hasattr(walk_list, "ok"):
    walk_list.ok = []
    walk_list.ko = []

  for value in value_list:
    if ( type( value ) is list or type( value ) is tuple ):
      walk_list( value )
    else:
      data_float = convert_to_float( value )
      if data_float:
        walk_list.ok.append( data_float )
      else:
        walk_list.ko.append( value )


  return [walk_list.ok, walk_list.ko]

walk_list( entry_list )
```

```python
import pyperclip
from numpy import random

ascii_range = {
    "signs"       : [[32,48],[58,65],[91,97],[123,127]],
    "numbers"     : [48,58],
    "upper_case"  : [65,91],
    "lower_case"  : [97,123]
}

def get_random_char( component ):

  if component == "signs":
    group = random.randint( 0, len(ascii_range["signs"]) )
    ascii_code = random.randint( ascii_range["signs"][group][0], ascii_range["signs"][group][1])
  else:
    ascii_code = random.randint( ascii_range[component][0], ascii_range[component][1])

  return chr( ascii_code )


def password_generator( long, upper_case = True, lower_case = True, numbers = True, signs = False, copy = False ):

  components = []
  password = []

  if upper_case:
    password.append( get_random_char( 'upper_case' ) )
    components.append( 'upper_case' )
  if lower_case:
    password.append( get_random_char( 'lower_case' ) )
    components.append( 'lower_case' )
  if numbers:
    password.append( get_random_char( 'numbers' ) )
    components.append( 'numbers' )
  if signs:
    password.append( get_random_char( 'signs') )
    components.append( 'signs' )


  for char in range( 0, long-len(password) ):
    component = random.randint( 0, len(components) )
    password.append( get_random_char(components[component]) )

  random.shuffle( password )
  str_password = "".join( password )

  if copy:
    pyperclip.copy( str_password )

  return str_password

password_generator( 10,  signs=True )
```

```python
def proces_game_line( game ):

  team1, team2        = game.strip().split('-')
  team1_name, goals1  = team1.rsplit("\t", 1)
  goals2, team2_name  = team2.split("\t", 1)

  tmp_board = safe_game_result(team1_name, int(goals1), int(goals2))
  tmp_board = safe_game_result(team2_name, int(goals2), int(goals1))

  return tmp_board

def get_points( gfor,gagainst ):
  return 1 if gfor == gagainst else ( 3 if gfor > gagainst else 0)

def safe_game_result( team, gfor, gagainst ):

  if not hasattr( safe_game_result, "games" ):
    safe_game_result.games = {}

  if safe_game_result.games.get(team) == None:
    points = get_points( gfor, gagainst )
    safe_game_result.games[team] = {
        'gfor'    : gfor,
        'gagainst': gagainst,
        'points'  : points,
        'wins'    : 1 if points == 3 else 0,
        'draws'   : 1 if points == 1 else 0,
        'losses'  : 1 if points == 0 else 0
    }
  else:
    points = get_points( gfor, gagainst )
    safe_game_result.games[team].update(
        gfor      = safe_game_result.games[team]["gfor"]    + gfor,
        gagainst  = safe_game_result.games[team]["gagainst"]+ gagainst,
        points    = safe_game_result.games[team]["points"]  + points,
        wins      = safe_game_result.games[team]["wins"]    + (1 if points == 3 else 0),
        draws     = safe_game_result.games[team]["draws"]   + (1 if points == 1 else 0),
        losses    = safe_game_result.games[team]["losses"]  + (1 if points == 0 else 0)
    )

  return safe_game_result.games

def get_data_required( data, field ):

  result = []
  for line in data:
    result.append( {'team': line[0], 'data':line[1][field]} )

  return result


def prepare_final_board( board ):

    b_gfor              = sorted( board.items(), key = lambda x: x[1]["gfor"], reverse=True )
    b_gagainst          = sorted( board.items(), key = lambda x: x[1]["gagainst"], reverse=True )
    b_classification    = sorted( board.items(), key = lambda x: x[1]["points"], reverse=True )

    b_final = {
        'Most goals for'    : get_data_required( b_gfor[:1], "gfor" ),
        'Most goals against': get_data_required( b_gagainst[:1], "gagainst" ),
        'Goals rating'      : get_data_required( b_gfor, "gfor" ),
        'Classification'    : get_data_required( b_classification, "points" )
    }

    return b_final


def process_games_file( name ):

  games = get_file( name, "list" )

  if games == None:
    return None

  for game in games:
    board = proces_game_line( game )

  return prepare_final_board( board )



process_games_file("historic partits.txt")

```
