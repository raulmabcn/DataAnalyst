##1. Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: 🔗 Índice de masa corporal IMC que es y como se calcula.

La funció ha de classificar el resultat en les seves respectives categories.



> Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.



````python
validation_limits = {
    'kg':{ 'min': 0, 'max': 300 },
    'hg':{ 'min': 0, 'max': 3 }
}

def convert_to_float( value ):

  try:
    data = float( value )
  except ValueError:
    data = False

  return data

def is_data_ok( unit, value ):
  return ( value < validation_limits[unit]['max'] ) and ( value > validation_limits[unit]['min'])

def get_user_data( field ):

  if field == 'kg':
    print("Introduce tu peso en kg (ej. 75.5): ")
  else:
    print("Introduce tu altura en metros (ej. 1.75): ")

  return convert_to_float( input() )


def request_data( field ):

  while True:
    data = get_user_data( field )
    if data and is_data_ok( field, data ):
      break
    else:
      print("Algo no fue bien, el valor no es correcto.")

  return data

def imc_calculator():

  checked_data = {}
  system_data = ['kg','hg']

  print("Calculadora de indice de masa corporal\n")

  for field in system_data:
      checked_data[field] = request_data( field )

  imc   = round( checked_data['kg'] / pow( checked_data['hg'], 2), 2 )

  if imc < 18.5:
    range = 'Bajo peso'
  elif imc < 25:
    range = 'Peso normal'
  elif imc < 30:
    range = 'Sobrepeso'
  else:
    range = 'Obesidad'

  print(f"Tu imc es: {imc}, se considera: {range} ")
````

##4. Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

Output esperat:

`reverse_dictionary({'a':1,'b':2,'c':3})`

`{1:'a',2:'b',3:'c'}`

`reverse_dictionary({'x':'apple', 'y':'banana','z':'banana'})`

`{'apple':'x','banana':['y','z']}`

````python
def reverse_diccionary( data ):

  if type(data) != dict:
    print("Error: los datos tiene que ser una lista!!")
    return None

  reversed_dict = {}
  duplicated_keys = []

  for key,value in data.items():

    if reversed_dict.get(value) == None:
      reversed_dict[value] = key
    elif isinstance( reversed_dict[value], list ):
       reversed_dict[value].append( key )
    else:
      reversed_dict[value] = [reversed_dict[value], key]
      duplicated_keys.append(value)

  if duplicated_keys:
    print(f"Atención! Los siguientes valores del diccionario original que no son únicos: {duplicated_keys}")

  return reversed_dict
````

````python
survey_1 = {'a':1,'b':2,'c':3}
survey_2 = {'x':'apple', 'y':'banana','z':'banana', 'g': 'apple'}

reverse_diccionary( survey_1 )
reverse_diccionary( survey_2 )
````

````python
def get_file( name, type ="text" ):
  try:
    with open(f"/content/{name}", "rt", encoding="utf-8") as f:
      return f.read() if type == "text" else f.readlines()
  except FileNotFoundError:
    print("No se encontro el fichero!")
    return None

def counter_sort_text_words( text ):

  if text == None:
    return None

  list_words = sorted(procces_phrase_to_list( clean_data( text ) ))

  words_counted = {}
  for word in list_words:
    words_counted[word] = 1 + words_counted.setdefault(word, 0)

  final_dict = {}
  for word, value in words_counted.items():
    if final_dict.get(word[0]) == None:
      final_dict[word[0]] = { word:value }
    else:
      final_dict[word[0]].update( {word:value} )

  return final_dict
````

````python
entry_list = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

def walk_list( value_list ):

  if not hasattr(walk_list, "ok"):
    walk_list.ok = []
    walk_list.ko = []

  for value in value_list:
    if ( type( value ) is list or type( value ) is tuple ):
      walk_list( value )
    else:
      data_float = convert_to_float( value )
      if data_float:
        walk_list.ok.append( data_float )
      else:
        walk_list.ko.append( value )


  return [walk_list.ok, walk_list.ko]
````

1. Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

    longitud (int): Longitud de la contrasenya
    majuscules (bool = True): Si hi ha d’aparèixer majúscules
    minuscules (bool = True): Si hi ha d’aparèixer minúscules
    numeros (bool = True): Si hi ha d’aparèixer números
    signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

Així doncs, si executem la funció de la següent manera:

`crear_contrasenya(10, True, True, True, True)`

Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

`9Er,5Vn8P$`

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.


> (EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).



````python
ascii_range = {
    "signs"       : [[32,48],[58,65],[91,97],[123,127]],
    "numbers"     : [48,58],
    "upper_case"  : [65,91],
    "lower_case"  : [97,123]
}

def get_random_char( component ):

  if component == "signs":
    group = random.randint( 0, len(ascii_range["signs"]) )
    ascii_code = random.randint( ascii_range["signs"][group][0], ascii_range["signs"][group][1])
  else:
    ascii_code = random.randint( ascii_range[component][0], ascii_range[component][1])

  return chr( ascii_code )


def password_generator( long, upper_case = True, lower_case = True, numbers = True, signs = False, copy = False ):

  components = []
  password = []

  if upper_case:
    password.append( get_random_char( 'upper_case' ) )
    components.append( 'upper_case' )
  if lower_case:
    password.append( get_random_char( 'lower_case' ) )
    components.append( 'lower_case' )
  if numbers:
    password.append( get_random_char( 'numbers' ) )
    components.append( 'numbers' )
  if signs:
    password.append( get_random_char( 'signs') )
    components.append( 'signs' )


  for char in range( 0, long-len(password) ):
    component = random.randint( 0, len(components) )
    password.append( get_random_char(components[component]) )

  random.shuffle( password )
  str_password = "".join( password )

  if copy:
    pyperclip.copy( str_password )

  return str_password
````

##2. Processament de dades simple

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:

    El nombre total de gols que ha fet cada equip.
    El nom de l’equip més golejador.
    El nom de l’equip més golejat
    La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)


````python
def get_points( gfor,gagainst ):
  return 1 if gfor == gagainst else ( 3 if gfor > gagainst else 0)

def get_data_required( data, field ):

  result = []
  for line in data:
    result.append( [line[0], line[1][field]] )

  return result

def prepare_final_board( board ):

    b_gfor              = sorted( board.items(), key = lambda x: x[1]["gfor"], reverse=True )
    b_gagainst          = sorted( board.items(), key = lambda x: x[1]["gagainst"], reverse=True )
    b_classification    = sorted( board.items(), key = lambda x: x[1]["points"], reverse=True )

    b_final = {
        'Most goals for'    : get_data_required( b_gfor[:1], "gfor" ),
        'Most goals against': get_data_required( b_gagainst[:1], "gagainst" ),
        'Goals rating'      : get_data_required( b_gfor, "gfor" ),
        'Classification'    : get_data_required( b_classification, "points" )
    }

    return b_final

def process_games_file( name ):

  games = get_file( name, "list" )

  if games == None:
    return None

  board = {}

  for game in games:
    team1_data, team2_data  = game.strip().split('-')
    team1, goals            = team1_data.rsplit("\t", 1)
    goals1 = int(goals)
    goals, team2            = team2_data.split("\t", 1)
    goals2 = int(goals)

    for team, gfor, gagainst in [[team1,goals1,goals2],[team2, goals2, goals1]]:

      if board.get(team) == None:
        points = get_points( gfor, gagainst )
        board[team] = {
            'gfor'    : gfor,
            'gagainst': gagainst,
            'points'  : points,
            'wins'    : 1 if points == 3 else 0,
            'draws'   : 1 if points == 1 else 0,
            'losses'  : 1 if points == 0 else 0
        }
      else:
        points = get_points( gfor, gagainst )
        board[team].update(
            gfor      = board[team]["gfor"]    + gfor,
            gagainst  = board[team]["gagainst"]+ gagainst,
            points    = board[team]["points"]  + points,
            wins      = board[team]["wins"]    + (1 if points == 3 else 0),
            draws     = board[team]["draws"]   + (1 if points == 1 else 0),
            losses    = board[team]["losses"]  + (1 if points == 0 else 0)
        )

  return prepare_final_board( board )
````

````python
game = process_games_file("historic partits.txt")

print(tabulate(game, headers='keys', tablefmt='simple_grid'))
````
