## Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades.

La funció ha de classificar el resultat en les seves respectives categories.

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
def request_correct_data( data_error ):

  user_input = {}
  message = {
      'kg': 'Introduce de nuevo tu peso en kg, siguiendo el formato (ej 80.3): ',
      'hg': 'Introduce de nuevo tu altura en metros, siguiendo el formato (ej 1.80) : '
  }

  print( f"Parece que hay un error en tus datos de entrada: { list( data_error.values() ) }\n" )

  for key,error in data_error.items():
    user_input[key] = input( f"{message[key]}" )

  return user_input

def convert_to_float(user_input):

  user_data   = {}
  data_error  = {}

  for data in user_input.keys():
    try:
      user_data[data] = float(user_input[data])

    except ValueError:
      data_error[data] = user_input[data]

  return user_data, data_error


def check_range_values( input, data_error ):

  check_data = {}
  range_values = { 'kg': [0,300], 'm': [0,3] }

  for key, value in input.items():
    if value <= range_values[key][0] or value >= range_values[key][1]:
      data_error[key]   = value
    else
      check_data[key]   = value

  return check_data, data_error


def check_data_values( user_input ):

  check_data = {}

  converted_input, data_error = convert_to_float( user_input )

  if converted_input:
    check_data, data_error = check_range_values( converted_input, data_error )

  return check_data, data_error



def get_filtered_data( user_input ):

  data_filtered, data_error = check_data_values( user_input )

  while data_error:
    user_input = request_correct_data( data_error )
    data_filtered_tmp, data_error = check_data_values( user_input )
    data_filtered.update( data_filtered_tmp )

  return data_filtered


def get_imc_range( user_imc ):

  if user_imc < 18.5:
    range = 'Bajo peso'
  elif user_imc < 25:
    range = 'Peso normal'
  elif user_imc < 30:
    range = 'Sobrepeso'
  else:
    range = 'Obesidad'

  return range


def print_message( user_data,user_imc ):

  kg    = user_data['kg']
  hg    = user_data['hg']
  range = get_imc_range( user_imc )

  print(f"Con tu peso {kg} kg y altura {hg} m,\ntu imc es: {user_imc}, el cual se considera {range}")


def calculate_imc( user_data):

  return round( user_data['kg'] / pow( user_data['hg'], 2), 2 )


def request_data():

  user_input = {}
  user_input['kg'] = input("Introduce tu peso en kg,  75.5: ")
  user_input['hg'] = input("\nIntroduce tu altura en metros, 1.75: ")

  return user_input


def main():

  print("Calculadora de indice de masa corporal\n")

  user_input  = request_data()
  user_data   = get_filtered_data( user_input )
  user_imc    = calculate_imc( user_data )

  print_message( user_data, user_imc )


main()





```
