## 1.Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades.

La funció ha de classificar el resultat en les seves respectives categories.

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
def convert_to_float( value ):

  try:
    data = float( value)
  except ValueError:
    data = False

  return data

def is_casting_ok( result, extra = None ):
  return result != False


def get_message_input_kg():

   print("Introduce tu peso en kg (ej. 75.5): ")

def get_message_input_hg():

  print("Introduce tu altura en metros (ej. 1.75): ")


def get_imc_range( user_imc ):

  if user_imc < 18.5:
    range = 'Bajo peso'
  elif user_imc < 25:
    range = 'Peso normal'
  elif user_imc < 30:
    range = 'Sobrepeso'
  else:
    range = 'Obesidad'

  return range

def get_message_error():

  print("Algo no fue bien, el valor no es correcto.")


validation_limits = {
    'kg':{ 'min': 0, 'max': 300 },
    'hg':{ 'min': 0, 'max': 3 }
}

def check_limits( unit, value ):

  return ( value < validation_limits[unit]['max'] ) and ( value > validation_limits[unit]['min'])



validations = {
    'cast_ok'   : is_casting_ok,
    'kg'        : lambda value : check_limits( 'kg', value ),
    'hg'        : lambda value : check_limits( 'hg', value )
}

requested_inputs = {
    'kg' : {
        "msg"       : get_message_input_kg,
        "msg_error" : get_message_error,
        "process"   : convert_to_float,
        "to_check"  : ['cast_ok','kg'],
        "checked_data" : False,
        "dummy"     : "67"
    },
    'hg'  :{
        "msg"       : get_message_input_hg,
        "msg_error" : get_message_error,
        "process"   : convert_to_float,
        "to_check"  : ['cast_ok','hg'],
        "checked_data" : False,
        "dummy"     : "1.65"
    },
}


def request_data( field, checked_data = None ):

  data_ok = False

  while True:

    if not requested_inputs[field]['checked_data']:
      requested_inputs[field]['msg']()
    else:
      requested_inputs[field]['msg']( checked_data )

    if dummy == True:
      print("Dummy Run!!!")
      data = requested_inputs[field]['dummy']
    else:
       data = input()

    casting_method = requested_inputs[field].get('process')
    if casting_method != None :
      data = casting_method( data )


    for check in requested_inputs[field]['to_check']:

        if not requested_inputs[field]['checked_data']:
          data_ok = validations[check](data)
        else:
          data_ok = validations[check](data, checked_data)

        if not data_ok:
          requested_inputs[field]['msg_error']()
          break

    if data_ok:
      break

  return data

def get_calculate_imc( user_data ):

  return round( user_data['kg'] / pow( user_data['hg'], 2), 2 )


def imc_calculator():

  checked_data = {}

  print("Calculadora de indice de masa corporal\n")

  for field in requested_inputs.keys():
      checked_data[field] = request_data( field )


  imc   = get_calculate_imc( checked_data )
  range = get_imc_range( imc )

  print(f"Tu imc es: {imc}, se considera: {range} ")


dummy = True
imc_calculator()





```

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

```python
list_convertions = {
    'C': {
        'name': 'Celcius',
        'min' : -273.15,
        'to'  : {
            'F'   : lambda x: (x * 1.8) + 32,
            'K'   : lambda x: x + 273.15,
            'Ra'  : lambda x: (x * 1.8) + 491.67,
            'Re'  : lambda x: x * 0.8
        }
    },
    'F': {
        'name': 'Fahrenheit',
        'min' : -459.67,
        'to'  : {
              'C'   : lambda x: (x - 32)/1.8,
              'K'   : lambda x: (x - 32) * 5/9 + 273.15,
              'Ra'  : lambda x: x + 459.67,
              'Re'  : lambda x: (x - 32)/2.25
        }
      },
    'K': {
        'name': 'Kelvin',
        'min' : 0,
        'to'  : {
              'C'  : lambda x: x - 273.15,
              'F'  : lambda x: (x - 273.15) * 1.8 + 32,
              'Ra' : lambda x: x * 1.8,
              'Re' : lambda x: (x - 273.15) * 0.8
          }
    },
    'Ra': {
        'name': 'Rankine',
        'min' : 0,
        'to'  : {
              'C'  : lambda x: (x - 491.67)/1.8,
              'F'  : lambda x: x - 459.67,
              'K'  : lambda x: x/1.8,
              'Re' : lambda x: (x - 491.67)/2.25
          }
    },
    'Re': {
        'name': 'Reamur',
        'min' : -218.52,
        'to'  : {
              'C'  : lambda x: x * 1.25,
              'F'  : lambda x: (x * 2.25) + 32,
              'K'  : lambda x: (x * 1.25) + 273.15,
              'Ra' : lambda x: (x * 2.25) + 491.67
          }
    }
}



def get_message_input_unit_from():

  print("Elige la unidad de origen de entre las de la lista introduciendo sus siglas (ej. C).")
  [print( f"{list_convertions[unit]['name']} ({unit}) ", end =" ") for unit in list_convertions.keys()]


def get_message_input_units_to( checked_data ):

  print("Elige la unidad/es de destino de entre las de la lista separadas por una coma (ej. F, C, Ra):")
  [print( f"{unit} ", end =" ") for unit in list_convertions[checked_data['unit_from']]['to'].keys()]


def get_message_input_temp( checked_data ):
  print(f"Introduce el valor de la temperatura, el mínimo es {list_convertions[checked_data['unit_from']]['min']}, (ej. 34.6):")

def convert_to_list( value ):

  return [unit.strip() for unit in value.split(",") if unit.strip()]


def validate_units_to_list( units, checked_data ):

  validate = [False]
  validate = [ None != list_convertions[checked_data['unit_from']]['to'].get(unit) for unit in units ]

  return all(validate)

validations = {
    'unit_from' : lambda unit : None != list_convertions.get(unit),
    'cast_ok'   : lambda value, default = None: value != False,
    'temp'      : lambda value, checked_data  : value >= list_convertions[checked_data['unit_from']]['min'],
    'units_to'  : validate_units_to_list
}

requested_inputs = {
    'unit_from' : {
        "msg"           : get_message_input_unit_from,
        "msg_error"     : get_message_error,
        "to_check"      : ['unit_from'],
        "checked_data"  : False,
        "dummy"         : "F"
    },
    'units_to'  :{
        "msg"           : get_message_input_units_to,
        "msg_error"     : get_message_error,
        "process"       : convert_to_list,
        "to_check"      : ['cast_ok','units_to'],
        "checked_data"  : True,
        "dummy"         : "C"
    },
    'temp'      :{
        "msg"           : get_message_input_temp,
        "msg_error"     : get_message_error,
        "process"       : convert_to_float,
        "to_check"      : ['cast_ok','temp'],
        "checked_data"  : True,
        "dummy"         : "3"
    },
}


def get_convert_temp(data):

    converted_temp = {
        'from'  : data['unit_from'],
        'value' : data['temp'],
        'to':{}
    }

    for unit_to in data['units_to']:
      converted_temp['to'][unit_to] = round( list_convertions[data['unit_from']]['to'][unit_to](data['temp']), 2)

    return converted_temp


def temp_conversor():

  checked_data = {}

  print("Conversor de temperaturas\n")

  for field in requested_inputs.keys():
      checked_data[field] = request_data( field, checked_data )


  return  get_convert_temp( checked_data )


dummy = True
result = temp_conversor()
print( result )
```
