````python
def calculate_age( birth_date, now ):

  birtday_pass = 1 if ( (birth_date.month, birth_date.day) >= (now.month, now.day) ) else 0

  return now.year - birth_date.year - birtday_pass

df['Data de Naixament'] = pd.to_datetime(df[['Any', 'Mes', 'Dia']].rename(
    columns={'Any': 'year', 'Mes': 'month', 'Dia': 'day'}
), errors='coerce')

now = pd.Timestamp.now()
df['Edad'] = df['Data de Naixament'].apply( calculate_age, args=(now,) )
````

````python
def request_start_city( cities ):

  print("¿Quieres que sea un recorrido circular? s/n")
  circular = input("\n")
  while circular not in ['s','n']:
    print("No te he entendido: s o n")
    circular = input("\n")

  is_circular = True if circular == 's' else False
  city_opt = cities.tolist()
  if not is_circular:
    city_opt.append('All')

  print("¿Desde que ciudad quieres empezar?:")
  print(f"{[city for city in city_opt]}")
  city = input("\n")

  while city not in city_opt:
    print("Esa ciudad no esta disponible, elige una de la lista")
    city = input()

  return city, is_circular
````

````python
def generate_coord_route( route, country = 'Spain' ):

  geolocator = Nominatim( user_agent="short route generator" )

  coord_route = []
  for city in route:
    try:
      localization = geolocator.geocode(f"{city}, {country}")
      if localization:
        coord_route.append( [localization.latitude, localization.longitude] )
        time.sleep(1)
      else:
        raise KeyError( "Ciudad no encontrada")
    except Exception as e:
      print( f"Algo paso: {e}" )

  return coord_route

def print_route_map( coord_route, route ):

  if coord_route:
    map = folium.Map( location=coord_route[0], zoom_start=6, tiles='cartodb positron')

    for i, city in enumerate(route):
      folium.Marker(
          location=coord_route[i],
          popup=f"Stop {i+1}: {city}"
      ).add_to(map)

    folium.PolyLine(
        coord_route,
        color='blue',
        weight=4,
        opacity=0.7
    ).add_to(map)

  return map
````

````python
# Primera iteración, prepara las páginas de ruta.
def generate_first_route_page( df_route, city_from, city_list ):

  route_page = []
  for city_to in city_list:
    route_page.append( [city_from, [city_to], df_route[city_from][city_to]] )

  return route_page

# Siguientes n iteraciones, calcula la distancia del recorrido propuesto.
def generate_n_route_page( df_route, route_page, city_from, city_list ):

  n_route_page = []
  for line in route_page:
    for city_to in city_list:
      if city_to in line[1]:
        continue
      cities =  line[1].copy()
      dist = df_route[cities[-1]][city_to]
      cities.append(city_to)

      n_route_page.append( [city_from, cities, line[2]+dist] )

  return n_route_page

def clean_route_page( n_route_page ):

  key_page = {}
  for line in n_route_page:
    # Indice para identificar, la ultima ciudad visitada, pasando por las mismas
    # ciudades independientemente del orden de visita.
    key_line = (line[1][-1], tuple(sorted(line[1])))

    # Si la clave coincide nos quedamos con la ruta de menor distancia.
    if key_line not in key_page or key_page[key_line][2] > line[2]:
      key_page[key_line] = line

  return list(key_page.values())

def calculate_best_route( city_from, is_circular ):

  df_route  = df.copy()
  city_list = df_route[city_from].index.drop(city_from)
  route_page = generate_first_route_page( df_route, city_from, city_list )

  # Mientras en la lista de ciudades visitadas no esten todas las ciudades de la
  # lista se sigue iterando
  while len(route_page[0][1]) < len(city_list):
    route_page  = generate_n_route_page( df_route, route_page, city_from, city_list )
    route_page  = clean_route_page( route_page )

  if is_circular:
    for index, line in enumerate( route_page ):
      route_page[index][2] += df_route[line[1][-1]][city_from]

  best_route_line = min( route_page, key=lambda line:line[2] )
  if is_circular:
    best_route_line[1].append(city_from)

  return best_route_line

def route_generator( ):

  df_cities  = df.columns

  start_city, is_circular = request_start_city( df_cities )

  start_city_list = [start_city] if start_city != 'All' else df_cities

  if start_city == 'All' and not is_circular:
    shortes_routes = []
    for city in df_cities:
      shortes_routes.append( calculate_best_route( city, is_circular ) )
    the_shorter_route = min( shortes_routes, key=lambda line:line[2] )
  else:
    the_shorter_route = calculate_best_route( start_city, is_circular )

  route = the_shorter_route[1].copy()
  route.insert(0, the_shorter_route[0])

  return route
````
