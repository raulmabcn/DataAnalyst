# EUT 2003, 2011 y 2024
## Carga y preparación esencial

Este notebook carga los tres ficheros armonizados, valida su estructura,
unifica tipos y categorías y genera el fichero intermedio utilizado por los
notebooks de análisis.

```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_DIR = Path.cwd().resolve()
if PROJECT_DIR.name == "Notebooks":
    PROJECT_DIR = PROJECT_DIR.parent
elif not (PROJECT_DIR / "CleanDataSets").exists():
    candidates = [parent for parent in PROJECT_DIR.parents if (parent / "CleanDataSets").exists()]
    if not candidates:
        raise FileNotFoundError("No se ha podido localizar la carpeta raiz V2.0")
    PROJECT_DIR = candidates[0]
COMMON_DIR = PROJECT_DIR / "PythonCommon"
if str(COMMON_DIR) not in sys.path:
    sys.path.insert(0, str(COMMON_DIR))

from eut_common import (
    DATA_FILES,
    PREPARED_DATA,
    load_harmonized_data,
    missing_summary,
    prepare_harmonized_data,
    save_prepared_data,
    validate_harmonized_datasets,
)
```

```python
pd.set_option("display.max_columns", 120)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

### Carga de los ficheros armonizados

```python
datasets = load_harmonized_data()

pd.DataFrame({
    "año": list(datasets),
    "fichero": [str(DATA_FILES[year]) for year in datasets],
    "filas": [len(data) for data in datasets.values()],
    "columnas": [data.shape[1] for data in datasets.values()],
})
```

### Validación estructural

```python
validate_harmonized_datasets(datasets)
print("Validaciones superadas para 2003, 2011 y 2024.")
```

### Unión y preparación

```python
eut = prepare_harmonized_data(datasets)
print(f"Filas totales: {len(eut):,}")
print(f"Columnas: {eut.shape[1]}")
eut.head()
```

### Diagnóstico de valores ausentes

```python
missing_overall, missing_by_year = missing_summary(eut)
display(missing_overall)
display(missing_by_year)
```

### Exportación para los notebooks de análisis

```python
output_path = save_prepared_data(eut)
print(f"Fichero preparado: {output_path}")
print(f"Dimensiones: {eut.shape[0]:,} filas x {eut.shape[1]} columnas")
```
